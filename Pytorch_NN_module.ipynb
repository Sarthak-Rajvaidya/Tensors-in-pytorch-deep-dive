```python
import torch
import torch.nn as nn

# Create model class
class Model(nn.Module):

    def __init__(self, num_features):  # Correct: __init__, not __int__
        super().__init__()

        self.linear1 = nn.Linear(num_features, 1)  # Correct: single =
        self.relu = nn.ReLU()
                       # Correct: single =
        self.linear2 = nn.Linear(3,1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):                  # Must be inside the class
        out = self.linear1(features)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)
        return out
```

```python
#create dataset
features = torch.rand(10,5)
#create model

model = Model(features.shape[1])

#Call the model for forward shape
model(features)

```
