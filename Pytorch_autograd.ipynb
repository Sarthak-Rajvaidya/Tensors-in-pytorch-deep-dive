```python
import torch
```

```python
x = torch.tensor(3.0,requires_grad=True) #requires_grad :- tesnor x ka derivate calcuate karne ke liye ..
```

```python
y = x**2
```

```python
x
```

```python
y
```

```python
y.backward()
```

```python
x.grad
```
